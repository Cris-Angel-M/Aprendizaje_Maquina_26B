```python
import numpy as np
import matplotlib.pyplot as plt
```

Paso 2. Modifica la siguiente clase para que tenga un método de entrenamiento con el algoritmo del perceptrón.

```python
class Perceptron:

  def __init__(self, n_inputs, learning_rate):
    self.w = - 1 + 2 * np.random.rand(n_inputs)
    self.b = - 1 + 2 * np.random.rand()
    self.eta = learning_rate

  def predict(self, X):
    _, p = X.shape
    y_est = np.zeros(p)
    for i in range(p):
      y_est[i] = np.dot(self.w, X[:,i])+self.b
      if y_est[i] >= 0:
        y_est[i]=1
      else:
        y_est[i]=0
    return y_est

  def fit(self, X, Y, epochs=50):
    _, p = X.shape    #_ es una variable anonima
    for _ in range(epochs):
      for i in range(p):
        # Escribe las ecuaciones del perceptrón
        y_est = self.predict(X[:,i].reshape(-1,1))
        self.w += self.eta*(Y[i]-y_est)*X[:,i]
        self.b += self.eta*(Y[i]-y_est)
```

Paso 3. Instancía la siguiente función para poder dibujar la línea que representa el Perceptrón.

```python
# Función para dibujar superficie de desición
def draw_2d_percep(model):
  w1, w2, b = model.w[0], model.w[1], model.b
  plt.plot([-2, 2],[(1/w2)*(-w1*(-2)-b),(1/w2)*(-w1*2-b)],'--k')
```

Paso 4. Corre el siguiente código para comprobar que la neurona es capaz de aprender la compuerta OR

```python
# Instanciar el modelo
model = Perceptron(2, 0.1)

# Datos
X = np.array([[0, 0, 1, 1],
              [0, 1, 0, 1]])
Y = np.array( [0, 1, 1, 1])

# Entrenar
model.fit(X,Y)

# Predicción
model.predict(X)

# Primero dibujemos los puntos
_, p = X.shape
for i in range(p):
  if Y[i] == 0:
    plt.plot(X[0,i],X[1,i], 'or')
  else:
    plt.plot(X[0,i],X[1,i], 'ob')

plt.title('Perceptrón OR')
plt.grid('on')
plt.xlim([-2,2])
plt.ylim([-2,2])
plt.xlabel(r'x1')
plt.ylabel(r'x2')

draw_2d_percep(model)
```

Respuesta: Algo característico del perceptrón es que está diseñado para trabajar únicamente con problemas linealmente separables. Si observamos los resultados de las compuertas OR y AND, con una sola línea recta es posible separar los puntos clasificados como ceros y unos de forma clara y rápida. Por el contrario, la compuerta XOR representa un problema no linealmente separable, por lo que una sola línea recta es incapaz de dividir correctamente las clases. Para poder clasificarlas de manera adecuada, se requeriría una frontera de decisión no lineal (una función que trace una curva) o bien el uso de una red neuronal multicapa que combine varias rectas.
